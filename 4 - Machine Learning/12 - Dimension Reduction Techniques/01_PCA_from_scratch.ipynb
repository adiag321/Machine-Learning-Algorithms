## Implementing PCA From Scratch

Link - https://www.machinelearningplus.com/machine-learning/principal-components-analysis-pca-better-explained/


```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
```

```python
url = 'https://raw.githubusercontent.com/selva86/datasets/master/mnist_012.csv'
df = pd.read_csv(url)
df.head()
```

```python
# Prepare X and Y
Y = df.loc[:, '0']
X = df.drop(['0'], axis=1)
```

```python
X_standard = X - X.mean()
X_standard
```

```python
df_cov = X_standard.cov()
print(df.shape)
df_cov.head()
```

```python
from numpy.linalg import eig
eigvalues, eigvectors = eig(df_cov)
print(eigvalues[:10])
print(eigvectors.shape)
```

```python
X_pca = np.dot(X_standard, eigvectors)
df_pca_calc = pd.DataFrame(X_pca) 
df_pca_calc.round(2).head()
```
